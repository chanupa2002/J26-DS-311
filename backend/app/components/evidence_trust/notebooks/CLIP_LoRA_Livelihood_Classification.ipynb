```python

import torch
from transformers import CLIPProcessor, CLIPModel

MODEL_NAME = "openai/clip-vit-base-patch32"
device = "cuda" if torch.cuda.is_available() else "cpu"

model = CLIPModel.from_pretrained(MODEL_NAME).to(device)
processor = CLIPProcessor.from_pretrained(MODEL_NAME)

model.eval()

print("Model loaded:", MODEL_NAME)
print("Device:", device)
```

```python

category_prompts = {
    "AGRICULTURE & PLANTATION CROPS":
        "A photo of crop cultivation, farming, or a plantation.",
    "LIVESTOCK & ANIMAL HUSBANDRY":
        "A photo of livestock farming, raising animals, or poultry farming.",
    "FOOD PROCESSING & PRODUCTION":
        "A photo of food preparation, baking, or food production.",
    "TRADE":
        "A photo of a retail shop, market stall, or goods being sold.",
    "GARMENT TEXTILE, HANDICRAFTS":
        "A photo of sewing, garment production, textile work, or handicrafts.",
    "CONSTRUCTION, BUILDING AND REPAIR":
        "A photo of building construction, masonry, or repair work.",
    "BEAUTY AND WELLNESS":
        "A photo of a beauty salon, hairdressing, skincare, or wellness service."
}

categories = list(category_prompts.keys())

print("Categories prepared:", len(categories))
for category in categories:
    print("-", category)
```

```python

from PIL import Image
import torch

sample = manifest.iloc[0]
image = Image.open(sample["filepath"]).convert("RGB")

texts = [category_prompts[c] for c in categories]

inputs = processor(
    text=texts,
    images=image,
    return_tensors="pt",
    padding=True
).to(device)

with torch.no_grad():
    outputs = model(**inputs)
    probabilities = outputs.logits_per_image.softmax(dim=1)[0]

ranked = sorted(
    zip(categories, probabilities.cpu().tolist()),
    key=lambda x: x[1],
    reverse=True
)

print("Image:", sample["filename"])
print("Actual category:", sample["category"])
print("\nCLIP predictions:")

for category, score in ranked:
    print(f"{category}: {score:.4f}")
```

```python

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from tqdm.auto import tqdm
import pandas as pd
import torch
from PIL import Image

results = []

for _, row in tqdm(manifest.iterrows(), total=len(manifest)):
    try:
        image = Image.open(row["filepath"]).convert("RGB")

        inputs = processor(
            text=[category_prompts[c] for c in categories],
            images=image,
            return_tensors="pt",
            padding=True
        ).to(device)

        with torch.no_grad():
            outputs = model(**inputs)
            scores = outputs.logits_per_image.softmax(dim=1)[0]

        predicted_index = scores.argmax().item()
        predicted_category = categories[predicted_index]

        results.append({
            "filename": row["filename"],
            "actual_category": row["category"],
            "predicted_category": predicted_category,
            "correct": predicted_category == row["category"]
        })

    except Exception as e:
        print(f"Could not process {row['filename']}: {e}")

baseline_results = pd.DataFrame(results)

print("Images evaluated:", len(baseline_results))
print("Baseline accuracy:", round(
    accuracy_score(
        baseline_results["actual_category"],
        baseline_results["predicted_category"]
    ), 4
))

print("\nClassification report:")
print(classification_report(
    baseline_results["actual_category"],
    baseline_results["predicted_category"],
    labels=categories,
    zero_division=0
))
```

```python
from PIL import Image
from collections import defaultdict
import hashlib
import os

def calculate_file_hash(filepath):
    """Calculate a hash of the file's contents."""
    hasher = hashlib.sha256()

    with open(filepath, "rb") as file:
        for chunk in iter(lambda: file.read(8192), b""):
            hasher.update(chunk)

    return hasher.hexdigest()


hash_to_files = defaultdict(list)

for _, row in training_manifest.iterrows():
    file_hash = calculate_file_hash(row["filepath"])
    hash_to_files[file_hash].append(row["filepath"])

duplicate_groups = [
    paths for paths in hash_to_files.values()
    if len(paths) > 1
]

print("Total images checked:", len(training_manifest))
print("Exact duplicate groups:", len(duplicate_groups))

for group_number, paths in enumerate(duplicate_groups, start=1):
    print(f"\nDuplicate group {group_number}:")
    for path in paths:
        print(" -", os.path.relpath(path, DATASET_PATH))
```

```python
!pip -q install ImageHash

import imagehash
from PIL import Image
from itertools import combinations

# Calculate perceptual hashes
image_hashes = {}

for _, row in training_manifest.iterrows():
    try:
        image = Image.open(row["filepath"]).convert("RGB")
        image_hashes[row["filename"]] = imagehash.phash(image)
    except Exception as e:
        print(f"Could not hash {row['filename']}: {e}")

# Find potentially similar image pairs
similar_pairs = []

for (name1, hash1), (name2, hash2) in combinations(
    image_hashes.items(), 2
):
    distance = hash1 - hash2

    # A small distance indicates visual similarity
    if distance <= 6:
        similar_pairs.append((name1, name2, distance))

print("Images hashed:", len(image_hashes))
print("Potential near-duplicate pairs:", len(similar_pairs))

for name1, name2, distance in similar_pairs:
    print(f"{name1} <-> {name2} | hash distance: {distance}")
```

```python
from sklearn.model_selection import train_test_split
import pandas as pd
import os

# Separate the underrepresented category
beauty_data = training_manifest[
    training_manifest["category"] == "BEAUTY AND WELLNESS"
].copy()

other_data = training_manifest[
    training_manifest["category"] != "BEAUTY AND WELLNESS"
].copy()

# Split the other six categories into 70% training and 30% temporary
train_other, temp_other = train_test_split(
    other_data,
    test_size=0.30,
    random_state=42,
    stratify=other_data["category"]
)

# Split temporary data equally into validation and test
val_other, test_other = train_test_split(
    temp_other,
    test_size=0.50,
    random_state=42,
    stratify=temp_other["category"]
)

# Keep all three Beauty & Wellness images in training
train_data = pd.concat(
    [train_other, beauty_data],
    ignore_index=True
)

val_data = val_other.reset_index(drop=True)
test_data = test_other.reset_index(drop=True)

# Add split labels for a saved manifest
train_data["split"] = "train"
val_data["split"] = "validation"
test_data["split"] = "test"

split_manifest = pd.concat(
    [train_data, val_data, test_data],
    ignore_index=True
)

split_path = os.path.join(
    DATASET_PATH,
    "seven_category_split_manifest.csv"
)

split_manifest.to_csv(split_path, index=False)

print("Training images:", len(train_data))
print("Validation images:", len(val_data))
print("Test images:", len(test_data))
print("Total:", len(split_manifest))

print("\nTraining distribution:")
display(train_data["category"].value_counts().sort_index().to_frame("count"))

print("\nValidation distribution:")
display(val_data["category"].value_counts().sort_index().to_frame("count"))

print("\nTest distribution:")
display(test_data["category"].value_counts().sort_index().to_frame("count"))

print("\nSaved split manifest:", split_path)
```

```python
import copy
import torch
import torch.nn as nn
from torch.optim import AdamW
from sklearn.metrics import accuracy_score, f1_score

# Confirm the model and classifier are on the same device
model.to(device)
classifier.to(device)

# Optimize only LoRA parameters and the classification head
trainable_params = [
    p for p in model.vision_model.parameters() if p.requires_grad
] + list(classifier.parameters())

optimizer = AdamW(trainable_params, lr=1e-4, weight_decay=0.01)
criterion = nn.CrossEntropyLoss()

# Extract image embeddings through CLIP's vision encoder
def get_image_embeddings(pixel_values):
    vision_output = model.vision_model(pixel_values=pixel_values)
    pooled_output = vision_output.pooler_output
    image_embeddings = model.visual_projection(pooled_output)
    return image_embeddings

# Quick forward-pass check using one training batch
model.eval()
classifier.eval()

sample_images, sample_labels = next(iter(train_loader))
sample_images = sample_images.to(device)

with torch.no_grad():
    sample_embeddings = get_image_embeddings(sample_images)
    sample_logits = classifier(sample_embeddings)

print("Sample embedding shape:", tuple(sample_embeddings.shape))
print("Sample logits shape:", tuple(sample_logits.shape))
print("Forward-pass check completed.")
```

```python

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt
import torch

# 1. Set both models to evaluation mode
model.eval()
classifier.eval()

all_labels = []
all_predictions = []

# 2. Evaluate without calculating gradients
with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        labels = labels.to(device).long()

        # Extract CLIP image embeddings
        embeddings = get_image_embeddings(images)

        # Predict one of the 7 classes
        logits = classifier(embeddings)
        predictions = torch.argmax(logits, dim=1)

        all_labels.extend(labels.cpu().tolist())
        all_predictions.extend(predictions.cpu().tolist())

# 3. Calculate overall metrics
accuracy = accuracy_score(all_labels, all_predictions)
macro_f1 = f1_score(
    all_labels, all_predictions,
    average="macro",
    zero_division=0
)
weighted_f1 = f1_score(
    all_labels, all_predictions,
    average="weighted",
    zero_division=0
)

print("===== TEST SET RESULTS =====")
print("Test images:", len(all_labels))
print(f"Accuracy:   {accuracy:.4f} ({accuracy * 100:.2f}%)")
print(f"Macro-F1:   {macro_f1:.4f}")
print(f"Weighted-F1: {weighted_f1:.4f}")

# 4. Detailed results for each class
print("\n===== CLASSIFICATION REPORT =====")
print(classification_report(
    all_labels,
    all_predictions,
    labels=list(range(7)),
    zero_division=0,
    digits=4
))

# 5. Confusion matrix
cm = confusion_matrix(
    all_labels,
    all_predictions,
    labels=list(range(7))
)

display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=list(range(7))
)

display.plot(values_format="d", xticks_rotation=45)
plt.title("Test Set Confusion Matrix (Class IDs)")
plt.tight_layout()
plt.show()
```

```python

import copy
import torch
from sklearn.metrics import accuracy_score, f1_score

# Calculate class weights from training labels only
class_counts = torch.zeros(7)

for _, labels in train_loader:
    for label in labels:
        class_counts[int(label)] += 1

class_weights = class_counts.sum() / (
    7 * class_counts.clamp(min=1)
)
class_weights = class_weights.to(device)

criterion = torch.nn.CrossEntropyLoss(weight=class_weights)

# Use your existing optimizer if it contains both:
# trainable LoRA parameters and classifier parameters.
best_val_loss = float("inf")
best_model_state = None
best_classifier_state = None
patience = 5
epochs_without_improvement = 0
num_epochs = 20

for epoch in range(num_epochs):

    # -------- TRAIN --------
    model.train()
    classifier.train()

    train_loss_sum = 0.0
    train_labels = []
    train_predictions = []

    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device).long()

        optimizer.zero_grad()

        embeddings = get_image_embeddings(images)
        logits = classifier(embeddings)
        loss = criterion(logits, labels)

        loss.backward()
        optimizer.step()

        train_loss_sum += loss.item() * labels.size(0)
        train_labels.extend(labels.cpu().tolist())
        train_predictions.extend(
            logits.argmax(dim=1).detach().cpu().tolist()
        )

    train_loss = train_loss_sum / len(train_labels)
    train_acc = accuracy_score(train_labels, train_predictions)

    # -------- VALIDATE --------
    model.eval()
    classifier.eval()

    val_loss_sum = 0.0
    val_labels = []
    val_predictions = []

    with torch.no_grad():
        for images, labels in val_loader:
            images = images.to(device)
            labels = labels.to(device).long()

            embeddings = get_image_embeddings(images)
            logits = classifier(embeddings)
            loss = criterion(logits, labels)

            val_loss_sum += loss.item() * labels.size(0)
            val_labels.extend(labels.cpu().tolist())
            val_predictions.extend(
                logits.argmax(dim=1).cpu().tolist()
            )

    val_loss = val_loss_sum / len(val_labels)
    val_acc = accuracy_score(val_labels, val_predictions)
    val_macro_f1 = f1_score(
        val_labels,
        val_predictions,
        labels=list(range(7)),
        average="macro",
        zero_division=0
    )

    print(
        f"Epoch {epoch + 1:02d}/{num_epochs} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.3f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.3f} | "
        f"Val Macro-F1: {val_macro_f1:.3f}"
    )

    # Keep the checkpoint with the lowest validation loss
    if val_loss < best_val_loss:
        best_val_loss = val_loss

        best_model_state = {
            k: v.detach().cpu().clone()
            for k, v in model.state_dict().items()
        }
        best_classifier_state = {
            k: v.detach().cpu().clone()
            for k, v in classifier.state_dict().items()
        }

        epochs_without_improvement = 0
    else:
        epochs_without_improvement += 1

    if epochs_without_improvement >= patience:
        print("Early stopping: validation loss did not improve.")
        break

# Restore the best validation checkpoint
if best_model_state is None:
    raise RuntimeError("No valid checkpoint was saved.")

model.load_state_dict(best_model_state)
classifier.load_state_dict(best_classifier_state)

model.eval()
classifier.eval()

print("\nTraining finished.")
print(f"Best validation loss: {best_val_loss:.4f}")
print("Best checkpoint restored.")
```

```python

import torch
import pandas as pd

results = []

model.eval()
classifier.eval()

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        labels = labels.to(device).long()

        embeddings = get_image_embeddings(images)
        logits = classifier(embeddings)

        probabilities = torch.softmax(logits, dim=1)
        confidences, predictions = probabilities.max(dim=1)

        for true_label, pred_label, confidence in zip(
            labels.cpu().tolist(),
            predictions.cpu().tolist(),
            confidences.cpu().tolist()
        ):
            results.append({
                "Actual_Class": true_label,
                "Predicted_Class": pred_label,
                "Confidence": round(confidence, 4),
                "Correct": true_label == pred_label
            })

results_df = pd.DataFrame(results)

print("Correct predictions:", results_df["Correct"].sum())
print("Incorrect predictions:", (~results_df["Correct"]).sum())

display(results_df[results_df["Correct"] == False])
```
